```python
print(torch.cuda.get_device_properties(0))
print(torch.version.cuda)
```

```python
props = torch.cuda.get_device_properties(0)

for attr in dir(props):
    if not attr.startswith("_"):
        value = getattr(props, attr)
        if not callable(value):
            print(f"{attr}: {value}")
```
